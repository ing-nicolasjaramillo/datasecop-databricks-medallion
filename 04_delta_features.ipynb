# Delta Lake: historial y recuperación de errores
Las tablas Delta guardan un registro de cada cambio. Eso permite consultar cómo estaba una tabla en el pasado y deshacer errores.

```python
%sql
USE CATALOG workspace_practica_1;
DESCRIBE HISTORY secop_silver.procesos;
```

## Simular un error
Imaginemos que alguien ejecuta por accidente una actualización que borra todos los precios.

```python
%sql
UPDATE secop_silver.procesos SET precio_base = 0;
SELECT SUM(precio_base) AS total_despues_del_error FROM secop_silver.procesos;
```

```python
%sql
DESCRIBE HISTORY secop_silver.procesos;
```

```python
%sql
-- Cambia el 1 por la versión anterior al UPDATE
SELECT SUM(precio_base) AS total_antes_del_error
FROM secop_silver.procesos VERSION AS OF 1;
```

%md
## Deshacer el error

```python
%sql
RESTORE TABLE secop_silver.procesos TO VERSION AS OF 1;
SELECT SUM(precio_base) AS total_recuperado FROM secop_silver.procesos;
```
